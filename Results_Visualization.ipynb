```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import numpy as np
import pandas as pd
```

# Train Loss

## Code + AST

```python
epochs = [1, 2, 3, 4, 5]
train_loss = [0.6179, 0.3821, 0.2545, 0.1854, 0.1289]
val_loss   = [0.2064, 0.1639, 0.1539, 0.1334, 0.1506]

plt.figure(figsize=(8, 6))

# Plot Training Loss (Blue)
plt.plot(epochs, train_loss, label='Training Loss', linestyle='-')

# Plot Validation Loss (Orange)
plt.plot(epochs, val_loss, label='Validation Loss', linestyle='-')

plt.title('Train Loss (Code + AST)')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(False)

plt.xticks(epochs)

plt.show()
```

## Code

```python
epochs = [1, 2, 3, 4, 5]
train_loss = [0.5711, 0.3752, 0.2250, 0.1524, 0.0702]
val_loss   = [0.2297, 0.1431, 0.1265, 0.1656, 0.1262]

plt.figure(figsize=(8, 6))

# Plot Training Loss (Blue)
plt.plot(epochs, train_loss, label='Training Loss', linestyle='-')

# Plot Validation Loss (Orange)
plt.plot(epochs, val_loss, label='Validation Loss', linestyle='-')

plt.title('Train Loss (Code)')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(False)

plt.xticks(epochs)

plt.show()
```

## Combination

```python
# Data
epochs = [1, 2, 3, 4, 5]
train_loss_code_ast = [0.6179, 0.3821, 0.2545, 0.1854, 0.1289]
val_loss_code_ast   = [0.2064, 0.1639, 0.1539, 0.1334, 0.1506]

train_loss_code = [0.5711, 0.3752, 0.2250, 0.1524, 0.0702]
val_loss_code   = [0.2297, 0.1431, 0.1265, 0.1656, 0.1262]

plt.style.use('seaborn-v0_8-whitegrid')

plt.figure(figsize=(10, 6))

# Colors
color_ast = '#2E86C1'
color_code = '#E74C3C'

# Plot "Code + AST"
plt.plot(epochs, train_loss_code_ast, label='Train (Code + AST)',
         color=color_ast, linestyle='-', linewidth=2.5, marker='o', markersize=6)
plt.plot(epochs, val_loss_code_ast, label='Val (Code + AST)',
         color=color_ast, linestyle='--', linewidth=2, marker='o', markersize=6, markerfacecolor='white')

# Plot "Code"
plt.plot(epochs, train_loss_code, label='Train (Code)',
         color=color_code, linestyle='-', linewidth=2.5, marker='s', markersize=6)
plt.plot(epochs, val_loss_code, label='Val (Code)',
         color=color_code, linestyle='--', linewidth=2, marker='s', markersize=6, markerfacecolor='white')

plt.title('Training vs. Validation Loss', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Epochs', fontsize=12, labelpad=10)
plt.ylabel('Loss', fontsize=12, labelpad=10)

plt.legend(frameon=True, fancybox=True, framealpha=0.9, loc='upper right', fontsize=10)

plt.grid(True, linestyle=':', alpha=0.6)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.xticks(epochs)
plt.gca().tick_params(axis='both', which='major', labelsize=10)

plt.tight_layout()
plt.show()
```

# Confusion Matrix

```python
total_human = 777
total_machine = 223

cm_ast_human_true = int(round(total_human * 0.42))
cm_ast_human_false = total_human - cm_ast_human_true
cm_ast_machine_true = int(round(total_machine * 0.73))
cm_ast_machine_false = total_machine - cm_ast_machine_true

cm_ast = np.array([[cm_ast_human_true, cm_ast_human_false],
                   [cm_ast_machine_false, cm_ast_machine_true]])

cm_code_human_true = int(round(total_human * 0.25))
cm_code_human_false = total_human - cm_code_human_true
cm_code_machine_true = int(round(total_machine * 0.89))
cm_code_machine_false = total_machine - cm_code_machine_true
cm_code = np.array([[cm_code_human_true, cm_code_human_false],
                    [cm_code_machine_false, cm_code_machine_true]])

labels = ['Human-Written', 'Machine-Generated']

# Plotting Confusion Matrices
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.heatmap(cm_ast, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=labels, yticklabels=labels, cbar=False, annot_kws={"size": 14})
axes[0].set_title('Confusion Matrix: Code + AST', fontsize=16)
axes[0].set_ylabel('Actual', fontsize=12)
axes[0].set_xlabel('Predicted', fontsize=12)

sns.heatmap(cm_code, annot=True, fmt='d', cmap='Reds', ax=axes[1],
            xticklabels=labels, yticklabels=labels, cbar=False, annot_kws={"size": 14})
axes[1].set_title('Confusion Matrix: Code Only', fontsize=16)
axes[1].set_ylabel('Actual', fontsize=12)
axes[1].set_xlabel('Predicted', fontsize=12)

plt.tight_layout()
plt.savefig('confusion_matrices.png')
```

```python
languages = ['C#', 'Go', 'Python', 'C++', 'Java', 'PHP', 'JavaScript', 'C']
counts = [122, 60, 303, 75, 256, 48, 85, 51]

lang_labels = [f"{l}\n(n={c})" for l, c in zip(languages, counts)]

f1_ast = [0.3604, 0.5217, 0.4904, 0.4167, 0.2081, 0.1429, 0.4516, 0.4828]
acc_ast = [0.4180, 0.8167, 0.4719, 0.6267, 0.4648, 0.2500, 0.4000, 0.7059]

f1_code = [0.3259, 0.5652, 0.4892, 0.3200, 0.2700, 0.1364, 0.5370, 0.4500]
acc_code = [0.2541, 0.6667, 0.4554, 0.3200, 0.3242, 0.2083, 0.4118, 0.5686]

x = np.arange(len(languages))
width = 0.35

fig2, ax2 = plt.subplots(figsize=(12, 6))
rects1 = ax2.bar(x - width/2, f1_ast, width, label='Code + AST', color='#2E86C1')
rects2 = ax2.bar(x + width/2, f1_code, width, label='Code Only', color='#E74C3C')

ax2.set_ylabel('F1 Score', fontsize=12)
ax2.set_title('F1 Score per Language', fontsize=16, fontweight='bold')
ax2.set_xticks(x)
ax2.set_xticklabels(lang_labels, fontsize=10)
ax2.legend()
ax2.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels
def autolabel(rects, ax):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9, rotation=0)

autolabel(rects1, ax2)
autolabel(rects2, ax2)

plt.tight_layout()
plt.savefig('language_f1_comparison.png')

# Plot Accuracy
fig3, ax3 = plt.subplots(figsize=(12, 6))
rects3 = ax3.bar(x - width/2, acc_ast, width, label='Code + AST', color='#2E86C1')
rects4 = ax3.bar(x + width/2, acc_code, width, label='Code Only', color='#E74C3C')

ax3.set_ylabel('Accuracy', fontsize=12)
ax3.set_title('Accuracy per Language', fontsize=16, fontweight='bold')
ax3.set_xticks(x)
ax3.set_xticklabels(lang_labels, fontsize=10)
ax3.legend()
ax3.grid(axis='y', linestyle='--', alpha=0.7)

autolabel(rects3, ax3)
autolabel(rects4, ax3)

plt.tight_layout()
plt.savefig('language_acc_comparison.png')
```
